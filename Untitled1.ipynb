```python
from collections import deque
graph = {
    'A':['B','C'],
    'B':[],
    'C':[],

}
visited = set()

queue = deque()
start = 'A'
queue.append(start)
print("BFS traversal:")
while queue:
    node = queue.popleft()
    if node not in visited:
        print(node,end=" ")
        visited.add(node)
        for neighbor in graph[node]:
            if neighbor not in visited:
                queue.append(neighbor)
```

```python
from collections import deque
graph = {
    'Origin': ['Edge1', 'Edge2'],
    'Edge1': ['Edge3', 'Edge4'],
    'Edge2': ['Edge5'],
    'Edge3': [],
    'Edge4': [],
    'Edge5': []
}

visited = set()
queue = deque()

start = 'Origin'

queue.append(start)

print("Cache invalidation propagation:")

while queue:
    server = queue.popleft()

    if server not in visited:
        print(server)
        visited.add(server)

        for neighbor in graph[server]:
            if neighbor not in visited:
                queue.append(neighbor)
```

```python
graph = {
    'A': ['B', 'C'],
    'B': [],
    'C': [],
}

visited = set()
stack = []
start = 'A'

stack.append(start)

print("DFS traversal:")

while stack:
    node = stack.pop()

    if node not in visited:
        print(node, end=" ")
        visited.add(node)
        for neighbor in reversed(graph[node]):
            if neighbor not in visited:
                stack.append(neighbor)
```
