```python
facts = [
    "Ram likes mango.",
    "Seema is a girl.",
    "Bill likes cindy.",
    "Rose is red.",
    "John owns gold."
]

print("Knowledge Base:")
for fact in facts:
    print(fact)

query = input("\nEnter a fact to check: ").strip().rstrip(".")

if query.lower() in [fact.rstrip(".").lower() for fact in facts]:
    print("Fact is true.")
else:
    print("Fact not found.")
```

```python
from kanren import Relation, facts, var, run, conde

parent = Relation()

facts(parent,
      ("Jhon", "David"),
      ("David", "Sam"),
      ("David", "Rita"))

def grandparent(x, z):
    y = var()
    return conde((parent(x, y), parent(y, z)))

x = var()

result = run(1, x, grandparent(x, "Sam"))
print("Grand parent of Sam is:", result[0])
```
